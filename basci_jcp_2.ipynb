## R3XA Tutorial 2 - Using predifined `settings` and `data_sources`
Saving, loading and using a predifined `setting` or `data_source`

```python
import r3xa_api as r3xa
```

**Creating a new data source** and **save it** in the bank of devices

```python
my_torque_sensor = r3xa.new_datasource.GenericDataSource(
    title='Dynamic Torque sensor',
    description='Andilog Dynamic Torque sensor. \nCapacity 6Nm. \nResolution : 0,6 mNm',
    output_components=1,
    output_dimension='point',
    output_units=[r3xa.unit('torque', 0, 'N.m'),],
    manufacturer='Andilog',
    model='SPIP TT 6',
    documentation='https://pdf.directindustry.com/pdf/andilog-technologies/andilog-force-torque-measurements/7999-34677.html'
    )
my_torque_sensor.save('devices/andilog_dynamic_torque_sensor.json')
```

**Loading an existing device** and add it to an existing R3XA File

```python
device = r3xa.DataSource.load('devices/andilog_dynamic_torque_sensor.json')
my_file = r3xa.R3XAFile.load('save/my_first_r3xa_file_v2.json')
my_file.data_sources.append(device)
print(my_file)
```
