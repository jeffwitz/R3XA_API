# Explore an existing R3XA document

This independent tutorial opens the existing torsion-test JSON. It does not require tutorials 0–5. Run both code cells to display the graph. The interactive graph is read-only: click a node to inspect its fields, or double-click a data node to focus on its neighbours.

```python
from pathlib import Path
from IPython.display import IFrame, display
from r3xa_api import R3XAFile

repo_root = next(
    (folder for folder in (Path.cwd(), *Path.cwd().parents) if (folder / 'pyproject.toml').is_file()),
    None,
)
if repo_root is None:
    raise RuntimeError('Start Jupyter from inside the R3XA_API repository.')
source = repo_root / 'tests' / 'fixtures' / 'documents' / 'essai-torsion.json'
document = R3XAFile.load(source)
document.validate()
print(f'Loaded: {document.title} ({len(document.data_sources)} sources, {len(document.data_sets)} sets)')
```

```python
output = repo_root / 'artifacts' / 'examples' / 'tutorials' / 'torsion_cytoscape.html'
output.parent.mkdir(parents=True, exist_ok=True)
html_path = document.plot(output, backend='cytoscape', include_description=False)
display(IFrame(
    src=f'/files/{html_path.relative_to(repo_root).as_posix()}',
    width='100%',
    height=760,
    extras=['sandbox="allow-scripts"'],
))
```

The iframe loads the standalone HTML from the notebook server. The file is saved at `artifacts/examples/tutorials/torsion_cytoscape.html`; it can also be opened separately in a browser.
